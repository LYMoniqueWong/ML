```python
import pandas as pd
df = pd.read_csv('heights.csv')
df.head()
```

###Detect outliers using percentile

```python
max_threshold = df.height.quantile(0.95)
max_threshold
```

```python
df[df.height>max_threshold]
```

```python
min_threshold = df.height.quantile(0.05)
min_threshold
```

```python
df[df.height<min_threshold]
```

###Remove outliers

```python
df[(df.height>min_threshold) & (df.height<max_threshold)]
```

###Another dataset: Air bnb NYC dataset

Remove outliers using percentile based on price per night for a given apartment/home.

We will use suitable upper and lower limits on percentile based on intuition. Our goal is to come up with a new pandas dataframe that doesn't have outliers.

```python
df = pd.read_csv('AB_NYC_2019.csv')
df.head()
```

```python
df.price.describe()
```

```python
min_threshold, max_threshold = df.price.quantile([0.01,0.999])
min_threshold, max_threshold
```

```python
df[df.price<min_threshold]
```

```python
df2 = df[(df.price>min_threshold) & (df.price<max_threshold)]
df2.shape
```

```python
df2.sample(5)
```

```python
df2.price.describe()
```
